Inspect the dataset structure, columns, data types, missing values, duplicates, and target distribution.

Add the executed analysis, outputs, and findings for this stage below.

```python

# ============================================================
# PROJECT: TELCO CUSTOMER CHURN CLASSIFICATION
# STAGE 1: DATA UNDERSTANDING
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("=" * 60)
print("TELCO CUSTOMER CHURN CLASSIFICATION")
print("STAGE 1: DATA UNDERSTANDING")
print("=" * 60)


# 2. LOAD DATASET
file_path = "../data/WA_Fn-UseC_-Telco-Customer-Churn.csv"

try:
    df = pd.read_csv(file_path)
    print("\nDataset successfully loaded!")
except FileNotFoundError:
    raise FileNotFoundError(
        "Dataset tidak ditemukan. Pastikan file CSV berada di folder data."
    )


# 3. DATASET PREVIEW
print("\n[1] DATASET PREVIEW")
display(df.head())


# 4. DATASET DIMENSIONS
print("\n[2] DATASET DIMENSIONS")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


# 5. COLUMN INFORMATION
print("\n[3] COLUMN INFORMATION")
print("Column names:")
print(df.columns.tolist())

print("\nData types and non-null values:")
df.info()


# 6. STATISTICAL SUMMARY
print("\n[4] STATISTICAL SUMMARY")
display(df.describe(include="all").T)


# 7. MISSING VALUE ANALYSIS
print("\n[5] MISSING VALUE ANALYSIS")

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_summary)


# 8. DUPLICATE DATA ANALYSIS
print("\n[6] DUPLICATE DATA ANALYSIS")

duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)


# 9. TARGET VARIABLE ANALYSIS
print("\n[7] TARGET VARIABLE ANALYSIS")

print("Churn distribution:")
display(df["Churn"].value_counts())

print("\nChurn percentage:")
display(
    (df["Churn"].value_counts(normalize=True) * 100)
    .round(2)
)


# 10. TARGET VARIABLE VISUALIZATION
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.savefig(
    "../images/churn_distribution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()


# 11. FINAL DATA SUMMARY
print("\n" + "=" * 60)
print("DATA UNDERSTANDING SUMMARY")
print("=" * 60)

print(f"Total Customers      : {df.shape[0]}")
print(f"Total Columns        : {df.shape[1]}")
print(f"Total Missing Values : {df.isnull().sum().sum()}")
print(f"Total Duplicate Rows : {duplicate_count}")
print(f"Target Variable      : Churn")
print(f"Target Classes       : {df['Churn'].nunique()}")

print("\nData understanding completed!")
```
