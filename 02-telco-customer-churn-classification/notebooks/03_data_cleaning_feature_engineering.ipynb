# 03 Data Cleaning Feature Engineering

Prepare the data, handle data quality issues, encode categorical variables, and prepare model features.

Add the executed analysis, outputs, and findings for this stage below.

```python

# ============================================================
# TELCO CUSTOMER CHURN CLASSIFICATION
# STAGE 3: DATA CLEANING & FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 60)
print("TELCO CUSTOMER CHURN CLASSIFICATION")
print("STAGE 3: DATA CLEANING & FEATURE ENGINEERING")
print("=" * 60)

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

data_path = "../data/WA_Fn-UseC_-Telco-Customer-Churn.csv"

df = pd.read_csv(data_path)

print("\nDataset successfully loaded!")
print(f"Initial shape: {df.shape}")

# ------------------------------------------------------------
# 2. CHECK INITIAL DATA CONDITION
# ------------------------------------------------------------

print("\n[1] INITIAL DATA CHECK")

print("\nMissing values before cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 3. DATA TYPE CONVERSION
# ------------------------------------------------------------

print("\n[2] DATA TYPE CONVERSION")

# Mengubah TotalCharges menjadi numerik
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("TotalCharges converted to numeric.")
print(df["TotalCharges"].dtype)

# ------------------------------------------------------------
# 4. HANDLE MISSING VALUES
# ------------------------------------------------------------

print("\n[3] HANDLE MISSING VALUES")

# Pada dataset Telco, pelanggan dengan tenure = 0
# dapat memiliki TotalCharges kosong karena belum
# memiliki riwayat tagihan.

zero_tenure = df["tenure"] == 0

df.loc[
    zero_tenure & df["TotalCharges"].isnull(),
    "TotalCharges"
] = 0

# Jika masih ada TotalCharges kosong pada pelanggan lain,
# isi dengan nilai median.
median_total = df["TotalCharges"].median()

df["TotalCharges"] = df["TotalCharges"].fillna(
    median_total
)

print("Missing values handled.")
print("\nRemaining missing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

# ------------------------------------------------------------
# 5. REMOVE DUPLICATE DATA
# ------------------------------------------------------------

print("\n[4] REMOVE DUPLICATES")

duplicates_before = df.duplicated().sum()

df = df.drop_duplicates()

print(f"Duplicates removed: {duplicates_before}")
print(f"Current shape: {df.shape}")

# ------------------------------------------------------------
# 6. FEATURE ENGINEERING
# ------------------------------------------------------------

print("\n[5] FEATURE ENGINEERING")

# Mengelompokkan lama berlangganan pelanggan
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 60, 72],
    labels=[
        "0-12 months",
        "13-24 months",
        "25-48 months",
        "49-60 months",
        "61-72 months"
    ]
)

print("New feature created: tenure_group")

print("\nTenure group distribution:")
print(df["tenure_group"].value_counts().sort_index())

# ------------------------------------------------------------
# 7. TARGET ENCODING
# ------------------------------------------------------------

print("\n[6] TARGET ENCODING")

# Mengubah target Churn:
# Yes = 1 (pelanggan berhenti)
# No  = 0 (pelanggan tetap)

df["Churn"] = df["Churn"].map({
    "Yes": 1,
    "No": 0
})

print("Churn target encoded.")
print(df["Churn"].value_counts())

# ------------------------------------------------------------
# 8. FINAL DATA VALIDATION
# ------------------------------------------------------------

print("\n[7] FINAL DATA VALIDATION")

print(f"Final dataset shape: {df.shape}")

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nTarget distribution:")
print(df["Churn"].value_counts())

# ------------------------------------------------------------
# 9. SAVE CLEANED DATASET
# ------------------------------------------------------------

output_path = Path("../data/cleaned_telco_churn.csv")

df.to_csv(output_path, index=False)

print("\n" + "=" * 60)
print("STAGE 3 COMPLETED SUCCESSFULLY!")
print("=" * 60)

print(f"Cleaned dataset saved to: {output_path}")
print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")
```
